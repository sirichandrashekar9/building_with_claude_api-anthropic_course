````python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-0"

def add_user_message(messages, text):
    message = {"role": "user", "content": text}
    messages.append(message)
    return messages

def add_assistant_message(messages, text):
    message = {"role": "assistant", "content": text}
    messages.append(message)
    return messages

def chat(messages, system=None, temperature=0, stop_sequences=[]):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature,
        "stop_sequences": stop_sequences
    }

    if system:
        params["system"] = system
        
    message = client.messages.create(**params)
    return message.content[0].text
````

````python
messages = []

#Prefilled assistant messages

add_user_message(messages, "Generate three different sample AWS CLI commands. Each should be very short")

add_assistant_message(messages, "```bash")

answer = chat(messages, stop_sequences=["```"])

answer.strip()
````
