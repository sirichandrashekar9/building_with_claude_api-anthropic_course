```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-0"

def add_user_message(messages, text):
    message = {"role": "user", "content": text}
    messages.append(message)
    return messages

def add_assistant_message(messages, text):
    message = {"role": "assistant", "content": text}
    messages.append(message)
    return messages

def chat(messages, system=None, temperature=1, stop_sequences=[]):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature,
        "stop_sequences": stop_sequences
    }

    if system:
        params["system"] = system
        
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

#Prefilled assistant messages

add_user_message(messages, "what is better harry potter or game of thrones?")

add_assistant_message(messages, "Harry Potter is better because")

answer = chat(messages)

answer
```

```python
messages = []

#stop sequence

add_user_message(messages, "List of all the states of United States in alphabetical order")

answer = chat(messages, stop_sequences=[" Hawaii"])
answer
```
